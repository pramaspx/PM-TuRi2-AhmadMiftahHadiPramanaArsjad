```python
import pandas as pd
import numpy as np

# Membuat dataset buatan sendiri
data = {
    'usia': [25, np.nan, 30, 35, np.nan, 40],
    'pendapatan': [5000000, 6000000, np.nan, 8000000, 7500000, np.nan],
    'pendidikan': ['SMA', 'S1', 'S2', 'SMA', 'S1', 'S2'],
    'kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Surabaya'],
    'membeli': ['Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak']

}

# Mengubah dictionary menjadi DataFrame
df = pd.DataFrame(data)

# Tampilkan data awal untuk memastikan
print(df)
```

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
print(df.isnull().sum(),)
print(df)
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
 X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
```

Latihan 4
Karena fit_transform digunakan pada data latih untuk mempelajari parameter statistik dan mengubah skalanya, sedangkan transform hanya diterapkan pada data uji menggunakan parameter yang sama persis dari data latih demi mencegah kebocoran data (data leakage).

Refleksi
1. untuk mencegah terjadinya data leakage (kebocoran data). Jika scaling dilakukan sebelum splitting, parameter statistik (seperti rata-rata atau standar deviasi) dari data uji akan ikut dihitung dan tercampur ke dalam data latih. Hal ini akan membuat performa model terlihat sangat bagus saat evaluasi, namun buruk ketika dihadapkan pada data baru di dunia nyata.

2.
- Label Encoding digunakan untuk variabel kategorikal yang bersifat ordinal
- One-Hot Encoding digunakan untuk variabel kategorikal yang bersifat nominal

3. Normalisasi mengubah rentang nilai fitur ke dalam skala tertentu, umumnya berada di antara nilai 0 dan 1. Sedangkan Standardisasi mengubah distribusi data agar memiliki nilai rata-rata (mean) sama dengan 0 dan standar deviasi sama dengan 1 (berpusat pada nol).
